```python

# ==========================================
# CODEALPHA INTERNSHIP - TASK 2
# AI BASED FAQ CHATBOT USING NLP
# ==========================================

# STEP 1: Install Libraries

!pip install nltk scikit-learn gradio -q


# STEP 2: Import Libraries

import nltk
import re
import gradio as gr

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('stopwords', quiet=True)

print("Libraries imported successfully!")


# STEP 3: Create FAQ Dataset

faqs = [

    {
        "question": "What is AI?",
        "keywords": "artificial intelligence smart machines",
        "answer": "AI stands for Artificial Intelligence. It enables machines to perform tasks that normally require human intelligence."
    },

    {
        "question": "What is Python?",
        "keywords": "python programming language coding",
        "answer": "Python is a high-level, interpreted programming language known for its simple syntax."
    },

    {
        "question": "What is Machine Learning?",
        "keywords": "ML learning data algorithms",
        "answer": "Machine Learning is a branch of AI that allows computers to learn from data and improve their performance."
    },

    {
        "question": "What is Deep Learning?",
        "keywords": "neural networks deep learning",
        "answer": "Deep Learning is a subset of Machine Learning that uses artificial neural networks to learn complex patterns."
    },

    {
        "question": "What is NLP?",
        "keywords": "natural language processing human language",
        "answer": "NLP stands for Natural Language Processing. It helps computers understand and process human language."
    },

    {
        "question": "What is a chatbot?",
        "keywords": "chat bot conversation virtual assistant",
        "answer": "A chatbot is a software application that communicates with users through text or voice."
    },

    {
        "question": "What is Data Science?",
        "keywords": "data analysis science statistics",
        "answer": "Data Science is a field that uses scientific methods and algorithms to extract useful information from data."
    },

    {
        "question": "What is an algorithm?",
        "keywords": "steps procedure problem solving",
        "answer": "An algorithm is a step-by-step procedure used to solve a particular problem."
    },

    {
        "question": "What is a neural network?",
        "keywords": "brain neurons deep learning",
        "answer": "A neural network is a computing model inspired by the human brain. It is used in Machine Learning."
    },

    {
        "question": "What is the difference between AI and ML?",
        "keywords": "artificial intelligence machine learning difference",
        "answer": "AI is the broader concept of making machines intelligent, while Machine Learning is a subset of AI that learns from data."
    },

    {
        "question": "Who created Python?",
        "keywords": "python developer creator",
        "answer": "Python was created by Guido van Rossum and was first released in 1991."
    },

    {
        "question": "What is supervised learning?",
        "keywords": "labeled data machine learning",
        "answer": "Supervised Learning is a type of Machine Learning where a model learns from labeled training data."
    },

    {
        "question": "What is unsupervised learning?",
        "keywords": "unlabeled data clustering",
        "answer": "Unsupervised Learning is a type of Machine Learning that finds patterns in unlabeled data."
    },

    {
        "question": "What is computer vision?",
        "keywords": "images videos visual recognition",
        "answer": "Computer Vision is a field of AI that enables computers to understand images and videos."
    },

    {
        "question": "What is an API?",
        "keywords": "application programming interface software",
        "answer": "API stands for Application Programming Interface. It allows different software applications to communicate."
    }

]


# STEP 4: NLP TEXT PREPROCESSING

stop_words = set(stopwords.words('english'))

stemmer = PorterStemmer()


def preprocess_text(text):

    text = text.lower()

    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.split()

    words = [
        stemmer.stem(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)


# STEP 5: Prepare FAQ Questions

faq_texts = []

for faq in faqs:

    combined_text = (
        faq["question"] + " " + faq["keywords"]
    )

    faq_texts.append(preprocess_text(combined_text))


# STEP 6: TF-IDF Vectorization

vectorizer = TfidfVectorizer(
    ngram_range=(1, 2)
)

faq_vectors = vectorizer.fit_transform(faq_texts)

print("FAQ Dataset processed successfully!")


# STEP 7: Chatbot Function Using Cosine Similarity

def chatbot(message, history):

    if not message.strip():

        return "Please enter a question!"

    if message.lower().strip() in ["bye", "goodbye", "exit"]:

        return "Thank you for using our AI FAQ Chatbot. Have a nice day!"

    if message.lower().strip() in ["hi", "hello", "hey"]:

        return "Hello! Welcome to our AI FAQ Chatbot. How can I help you?"

    # Preprocess user question

    processed_message = preprocess_text(message)

    # Convert user question into TF-IDF vector

    user_vector = vectorizer.transform([processed_message])

    # Calculate cosine similarity

    similarity_scores = cosine_similarity(
        user_vector,
        faq_vectors
    )

    # Find the best matching FAQ

    best_match = similarity_scores.argmax()

    best_score = similarity_scores[0][best_match]

    # Set similarity threshold

    if best_score < 0.12:

        return "Sorry, I couldn't understand your question. Please ask something related to AI or Python."

    # Return matching answer

    answer = faqs[best_match]["answer"]

    return answer


# STEP 8: Create Gradio Chat Interface

demo = gr.ChatInterface(

    fn=chatbot,

    title="🤖 AI-Based FAQ Chatbot",

    description=(
        "Welcome! Ask me anything about "
        "Artificial Intelligence, Python, "
        "Machine Learning and Deep Learning."
    ),

    examples=[
        "What is AI?",
        "What is Python?",
        "What is Machine Learning?",
        "What is Deep Learning?",
        "What is NLP?",
        "What is a chatbot?"
    ]

)


# STEP 9: Launch Chatbot

print("AI FAQ Chatbot is ready!")

demo.launch(share=True)
```
